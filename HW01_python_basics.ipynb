````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
# 1
cities_clean = [c.strip().title() for c in cities]  # title() преобразует первую букву каждого слова в строке к верхнему регистру
print(cities_clean)

# 2
amounts_int = [int(i) for i in amounts]
print(amounts_int)

# 3
total_amount = len(amounts_int)
total_sum = sum(amounts_int)
avg_check = round(total_sum / total_amount)
print(f"заказов: {total_amount}, сумма: {total_sum}, средний чек: {avg_check}")

# 4
unpaid_count = paid.count(False)
unpaid_sum = sum(a for a, p in zip(amounts_int, paid) if not p)
print(f"не оплачено: {unpaid_count} заказа на {unpaid_sum}")

# 5
print(f"{'город':<10} | {'сумма':>5} | статус")
for c, a, p in zip(cities_clean, amounts_int, paid):
    if p:
        status = "оплачен" 
    else: 
        status ="не оплачен"
    print(f"{c:<10} | {a:>5} | {status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
# 6
almaty_rev = sum(a for c, a, p in zip(cities_clean, amounts_int, paid) if p and c == "Almaty")
astana_rev = sum(a for c, a, p in zip(cities_clean, amounts_int, paid) if p and c == "Astana")
print(f"Алматы: {almaty_rev}, Астана: {astana_rev}")

# 7
max_amount = 0   
max_index = -1   
for i in range(len(amounts_int)):      
    if paid[i] == True:               
        if amounts_int[i] > max_amount:   
            max_amount = amounts_int[i]   
            max_index = i                
print("Индекс:", max_index , "Город:", cities_clean[max_index], "Сумма:", max_amount)

# 8
weekend = sum(1 for d in days if d in ("сб", "вс"))
percent = weekend / len(days) * 100
print(f"Выходные: {weekend} из {len(days)} = {percent:.1f}%")

# 9
goal = 400000
saved, months, deposit = 0, 0, 15000
while saved < goal:
    saved += deposit
    months += 1
print(f"за: {months} мес., накопит: {saved}")

# 9 - если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?
saved, months, deposit = 0, 0, 15000
while saved < goal:
    saved += deposit
    months += 1
    deposit += 1000
print(f"за: {months} мес., накопит: {saved}")    
````
